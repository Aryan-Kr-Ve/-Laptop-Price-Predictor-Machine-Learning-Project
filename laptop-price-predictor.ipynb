```python
# 1. Standardize text and units
df['Memory'] = df['Memory'].astype(str).str.replace(r'\.0', '', regex=True)
df['Memory'] = df['Memory'].str.replace('GB', '', regex=False)
df['Memory'] = df['Memory'].str.replace('TB', '000', regex=False)

# 2. Split dual storage configurations (e.g., "128 SSD + 1000 HDD")
new = df['Memory'].str.split('+', n=1, expand=True)

df['first'] = new[0].str.strip()
df['second'] = new[1].fillna('0').str.strip()

# 3. Detect storage types for both slots
for prefix, col in [('Layer1', 'first'), ('Layer2', 'second')]:
    df[f'{prefix}HDD'] = df[col].str.contains('HDD').astype(int)
    df[f'{prefix}SSD'] = df[col].str.contains('SSD').astype(int)
    df[f'{prefix}Hybrid'] = df[col].str.contains('Hybrid').astype(int)
    df[f'{prefix}Flash_Storage'] = df[col].str.contains('Flash Storage').astype(int)

# 4. Extract numeric capacities (explicit regex=True)
df['first'] = df['first'].str.replace(r'\D', '', regex=True).astype(int)
df['second'] = df['second'].str.replace(r'\D', '', regex=True).astype(int)

# 5. Compute total storage per category
df['HDD'] = (df['first'] * df['Layer1HDD']) + (df['second'] * df['Layer2HDD'])
df['SSD'] = (df['first'] * df['Layer1SSD']) + (df['second'] * df['Layer2SSD'])
df['Hybrid'] = (df['first'] * df['Layer1Hybrid']) + (df['second'] * df['Layer2Hybrid'])
df['Flash_Storage'] = (df['first'] * df['Layer1Flash_Storage']) + (df['second'] * df['Layer2Flash_Storage'])

# 6. Drop intermediate columns
intermediate_cols = [
    'first', 'second',
    'Layer1HDD', 'Layer1SSD', 'Layer1Hybrid', 'Layer1Flash_Storage',
    'Layer2HDD', 'Layer2SSD', 'Layer2Hybrid', 'Layer2Flash_Storage'
]
df.drop(columns=intermediate_cols, inplace=True)
```

### Linear regression

```python
# Step 1: Preprocessing with OneHotEncoder
step1 = ColumnTransformer(
    transformers=[
        (
            'col_tnf',
            OneHotEncoder(
                sparse_output=False, drop='first', handle_unknown='ignore'
            ),
            [0, 1, 7, 10, 11],
        )
    ],
    remainder='passthrough',
)

# Step 2: Model
step2 = LinearRegression()

# Pipeline construction
pipe = Pipeline([('step1', step1), ('step2', step2)])

# Training and evaluation
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)

print('R2 score:', r2_score(y_test, y_pred))
print('MAE:', mean_absolute_error(y_test, y_pred))
```

### Voting Regressor

```python
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import (
    ExtraTreesRegressor,
    GradientBoostingRegressor,
    RandomForestRegressor,
    VotingRegressor,
)
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBRegressor

# Step 1: Preprocess categorical features
step1 = ColumnTransformer(
    transformers=[
        (
            'col_tnf',
            OneHotEncoder(
                sparse_output=False,
                drop='first',
                handle_unknown='infrequent_if_exist',
            ),
            [0, 1, 7, 10, 11],
        )
    ],
    remainder='passthrough',
)

# Step 2: Define base ensemble models
rf = RandomForestRegressor(
    n_estimators=350,
    random_state=3,
    max_samples=0.5,
    max_features=0.75,
    max_depth=15,
)
gbdt = GradientBoostingRegressor(
    n_estimators=100, max_features=0.5, random_state=3
)
xgb = XGBRegressor(
    n_estimators=25, learning_rate=0.3, max_depth=5, random_state=3
)
# Note: bootstrap=True is required when using max_samples in ExtraTrees
et = ExtraTreesRegressor(
    n_estimators=100,
    random_state=3,
    bootstrap=True,
    max_samples=0.5,
    max_features=0.75,
    max_depth=10,
)

# Step 3: Combine with VotingRegressor
step2 = VotingRegressor(
    estimators=[('rf', rf), ('gbdt', gbdt), ('xgb', xgb), ('et', et)],
    weights=[5, 1, 1, 1],
)

# Build and execute pipeline
pipe = Pipeline([('step1', step1), ('step2', step2)])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)

print('R2 score:', r2_score(y_test, y_pred))
print('MAE:', mean_absolute_error(y_test, y_pred))
```

### Stacking

```python
from sklearn.ensemble import StackingRegressor
# Step 1: Preprocess categorical columns
step1 = ColumnTransformer(
    transformers=[
        (
            'col_tnf',
            OneHotEncoder(
                sparse_output=False,
                drop='first',
                handle_unknown='infrequent_if_exist',
            ),
            [0, 1, 7, 10, 11],
        )
    ],
    remainder='passthrough',
)

# Step 2: Base estimators for stacking
estimators = [
    (
        'rf',
        RandomForestRegressor(
            n_estimators=350,
            random_state=3,
            max_samples=0.5,
            max_features=0.75,
            max_depth=15,
        ),
    ),
    (
        'gbdt',
        GradientBoostingRegressor(
            n_estimators=100, max_features=0.5, random_state=3
        ),
    ),
    (
        'xgb',
        XGBRegressor(
            n_estimators=25, learning_rate=0.3, max_depth=5, random_state=3
        ),
    ),
]

# Step 3: Meta-model / Stacking Regressor
step2 = StackingRegressor(
    estimators=estimators,
    final_estimator=Ridge(alpha=100.0),
    cv=5,  # 5-fold cross-validation to generate out-of-fold meta-features
    n_jobs=-1,  # Parallelize base estimator training
)

# Build & execute Pipeline
pipe = Pipeline([('step1', step1), ('step2', step2)])

pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)

print('R2 score:', r2_score(y_test, y_pred))
print('MAE:', mean_absolute_error(y_test, y_pred))
```
